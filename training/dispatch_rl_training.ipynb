```python
# Cell 3: Configure training parameters
TASK  = 'easy'    # easy | medium | hard | crisis
STEPS = 300       # training steps (reduce to 50 for a quick smoke test)
EPISODES = 200    # prompt dataset size
print(f'Config: task={TASK}, steps={STEPS}, episodes={EPISODES}')
```

```python
# Cell 4: Run training
!python training/train_dispatch.py \
    --task {TASK} \
    --steps {STEPS} \
    --episodes {EPISODES}
```

```python
# Cell 5: Display reward curve
from IPython.display import Image, display
display(Image('training/reward_curve.png'))
```

```python
# Cell 6: Print before/after comparison
import json
with open('training/comparison.json') as f:
    c = json.load(f)
print('=' * 50)
print(f"Task:       {c['task']}")
print(f"Steps:      {c['steps']}")
print(f"Baseline:   {c['baseline_score']:.4f}  (random agent)")
print(f"Trained:    {c['trained_score']:.4f}  (GRPO)")
print(f"Improvement: {c['improvement']:+.4f}")
print('=' * 50)
```
